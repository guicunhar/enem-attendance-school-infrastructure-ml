```python
description = data_set['PRESENCA_COMPLETA'].describe()
print(description)
```

```python
low = ((data_set["PRESENCA_COMPLETA"] <= 0.70)).sum()
medium = ((data_set["PRESENCA_COMPLETA"] > 0.70) & (data_set["PRESENCA_COMPLETA"] <= 0.88)).sum()  # 70% to 88%
high = (data_set["PRESENCA_COMPLETA"] > 0.88).sum()        # mean above 88%

total_schools = len(data_set)
percent_low = (low / total_schools) * 100
percent_medium = (medium / total_schools) * 100
percent_high = (high / total_schools) * 100

print(f"{percent_low:.2f}% of schools have LOW mean attendance")
print(f"{percent_medium:.2f}% of schools have MEDIUM mean attendance")
print(f"{percent_high:.2f}% of schools have HIGH mean attendance")
```

```python
data_set = data_set.fillna(0)
print("Number of null values after filling all with 0:")
print(data_set.isnull().sum().sum())
```

# Handling Regions

```python
# Convert to category type first
data_set['NO_REGIAO'] = data_set['NO_REGIAO'].astype('category')
data_set['SG_UF'] = data_set['SG_UF'].astype('category')

display(data_set.head())
```

```python
# Create a mapping from region to a starting number
region_map = {region: i * 100 for i, region in enumerate(data_set['NO_REGIAO'].unique())}

# Create a mapping for UF codes based on region
uf_map = {}
for region, start_code in region_map.items():
    ufs_in_region = data_set[data_set['NO_REGIAO'] == region]['SG_UF'].unique()
    for i, uf in enumerate(ufs_in_region):
        uf_map[uf] = start_code + i

# Apply the custom UF encoding
data_set['SG_UF_CODED'] = data_set['SG_UF'].map(uf_map)
data_set['NO_REGIAO_CODED'] = data_set['NO_REGIAO'].map(region_map)


# Drop the original SG_UF and NO_REGIAO columns
data_set = data_set.drop(columns=['SG_UF', 'NO_REGIAO'])

display(data_set[['NO_REGIAO_CODED', 'SG_UF_CODED']].head())
```

# Handling Outliers

```python
# Remove a school with too many students
# Distance-learning (EAD) school with 35000+ students
# print(data_set[data_set['CO_ESCOLA'] == 41156200])
data_set = data_set[data_set['QT_MAT_PROF'] <= 20000].copy()
```

# COMBINING FEATURES

```python
# Combine columns
data_set['IN_PATIO'] = np.where((data_set['IN_PATIO_COBERTO'] == 1) | (data_set['IN_PATIO_DESCOBERTO'] == 1), 1, 0)
data_set['IN_ARTES'] = np.where((data_set['IN_SALA_ATELIE_ARTES'] == 1) | (data_set['IN_SALA_MUSICA_CORAL'] == 1) | (data_set['IN_SALA_ESTUDIO_DANCA'] == 1), 1, 0)
# data_set['IN_TODOS_BASICOS'] = np.where((data_set['IN_AGUA'] == 1) & (data_set['IN_ENERGIA'] == 1) & (data_set['IN_ESGOTO'] == 1) & (data_set['IN_TRATAMENTO_LIXO'] == 1), 1, 0)
data_set['TOTAL_BASICOS'] = data_set['IN_AGUA'] + data_set['IN_ENERGIA'] + data_set['IN_ESGOTO'] + data_set['IN_TRATAMENTO_LIXO']
data_set['IN_BASICOS_TEC'] = np.where((data_set['IN_EQUIP_IMPRESSORA'] == 1) & (data_set['IN_EQUIP_SOM'] == 1) & (data_set['IN_EQUIP_TV'] == 1) & (data_set['IN_EQUIP_MULTIMIDIA'] == 1), 1, 0)
data_set['IN_COMP_ALUNO'] = np.where((data_set['IN_DESKTOP_ALUNO'] == 1) | (data_set['IN_COMP_PORTATIL_ALUNO'] == 1), 1, 0)

# Remove columns
columns_to_remove = [
    'IN_PATIO_COBERTO',
    'IN_PATIO_DESCOBERTO',
    'IN_SALA_ATELIE_ARTES',
    'IN_SALA_MUSICA_CORAL',
    'IN_SALA_ESTUDIO_DANCA',
    'IN_SALA_MULTIUSO',
    'IN_SALA_ESTUDIO_GRAVACAO',
    'IN_SALA_OFICINAS_EDUC_PROF',
    'IN_SECRETARIA',
    'IN_AGUA',
    'IN_ENERGIA',
    'IN_ESGOTO',
    'IN_TRATAMENTO_LIXO',
    'IN_EQUIP_IMPRESSORA',
    'IN_EQUIP_SOM',
    'IN_EQUIP_TV',
    'IN_EQUIP_MULTIMIDIA',
    'IN_DESKTOP_ALUNO',
    'IN_COMP_PORTATIL_ALUNO',
    'IN_ALIMENTACAO',
    'IN_EJA',
]

data_set = data_set.drop(columns=columns_to_remove)

# Rename column
data_set = data_set.rename(columns={'IN_RESERVAA': 'IN_COTAS'})

display(data_set.head())
```

```python
# Concatenate the string values of 'VINCULO_PUBLICO' and 'VINCULO_PRIVADO'
data_set['VINCULO'] = data_set['VINCULO_PUBLICO'] + data_set['VINCULO_PRIVADO']

# Display the relevant columns to verify
display(data_set[['VINCULO_PUBLICO', 'VINCULO_PRIVADO', 'VINCULO']].head())

DROP_C =['VINCULO_PUBLICO', 'VINCULO_PRIVADO']
data_set = data_set.drop(columns=DROP_C)
```
